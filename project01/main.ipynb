```python
products = df.columns[24:48]
products
```

```python
dates = df.groupby('fecha_dato')[['ncodpers']].sum()
dates=dates.index
dates
```

```python
names = {}
for product in products:
    names[product] = product + '_prev'
names
```

```python
result = []


for i in range(1, len(dates)):

    # 전번달
    previous = df.loc[
        df['fecha_dato'] == dates[i - 1]
    ].set_index('ncodpers')[products]

    # 이번달
    current = df.loc[
        df['fecha_dato'] == dates[i]
    ].set_index('ncodpers')[products]

    previous = previous.rename(columns=names)

    # 연결
    compared = current.join(previous)

    for product in products:
        new_count = (
            (compared[product + '_prev'] == 0)
            & (compared[product] == 1)
        ).sum()

        result.append({
            'fecha_dato': dates[i],
            '상품': product,
            '신규가입건수': new_count
        })
result
```

```python
monthly_result = pd.DataFrame(result)
monthly_result
```

```python
total_result = monthly_result.groupby('상품')[['신규가입건수']].sum()
total_result = total_result.sort_values(by='신규가입건수', ascending=False)
#total_result
total_result.iloc[:5]
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
```

```python
columns = [
    'sexo', 'age', 'segmento', 'ind_empleado', 'tiprel_1mes'
] + list(products)

last = df.loc[df['fecha_dato'] == '2016-05-28', columns]
#last
#triple_1mes에 NaN값 존재
last = last.dropna()
```

```python
last['상품수'] = last[products].sum(axis=1)
last['가입여부'] = 0
last.loc[last['상품수'] >= 3, '가입여부'] = 1
```

```python
last.loc[last['age'] < 30, '연령대'] = 'young'
last.loc[last['age'] >= 30, '연령대'] = 'adult'
last.loc[last['age'] >= 65, '연령대'] = 'senior'
```

```python
criteria = [
    'sexo', '연령대', 'segmento', 'ind_empleado', 'tiprel_1mes'
]

y = last['가입여부']

results = []

for criterion in criteria:

    X = pd.get_dummies(last[criterion])

    #점수 계산
    bestfeatures = SelectKBest(score_func=chi2, k='all')
    fit = bestfeatures.fit(X, y)

    score = fit.scores_.sum()

    results.append({
        '기준': criterion,
        '카이제곱': score
    })

result = pd.DataFrame(results)
result = result.sort_values(by='카이제곱', ascending=False)

result
```
